## Setup

```python
import numpy as np
import matplotlib.pyplot as plt

import sionna.rt as rt
from sionna.rt import (load_scene, scene as scene_assets,
                       Transmitter, Receiver, PlanarArray,
                       PathSolver, RadioMapSolver, Camera, RadioMaterial)

# the central Munich urban scene used in many chapters of this book
scene = load_scene(scene_assets.munich)
scene.frequency = 3.5e9
scene.tx_array = PlanarArray(num_rows=1, num_cols=1,
                             pattern='iso', polarization='V')
scene.rx_array = PlanarArray(num_rows=1, num_cols=1,
                             pattern='iso', polarization='V')
scene.add(Transmitter('tx0', position=[-30.0, 38.0, 35.0]))

solver = PathSolver()
rmsolver = RadioMapSolver()
```

## 5. Setting the Scattering Coefficient

```python
from sionna.rt import RadioMaterial

mat = RadioMaterial(name='rough_concrete',
                     relative_permittivity=5.31,
                     conductivity=0.052,
                     scattering_coefficient=0.5,
                     xpd_coefficient=0.0)
```

```python
# set the scattering coefficient on every material in the scene
for mat in scene.radio_materials.values():
    mat.scattering_coefficient = 0.5
```

```python
rm_scat = rmsolver(scene, max_depth=3, diffuse_reflection=True,
                   cell_size=[2.0,2.0], samples_per_tx=10**6)
```
