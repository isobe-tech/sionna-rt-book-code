## 6. GPU Execution and CPU Execution

````python
import sionna.rt as rt   # the variant is selected here
import drjit as dr
import mitsuba as mi

cuda_ok = dr.has_backend(dr.JitBackend.CUDA)
llvm_ok = dr.has_backend(dr.JitBackend.LLVM)
print("Active Mitsuba variant:", mi.variant())
print("GPU (CUDA) available:", cuda_ok)
print("CPU (LLVM) available:", llvm_ok)
````
