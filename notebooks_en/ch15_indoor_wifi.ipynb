## Setup

```python
import numpy as np
import matplotlib.pyplot as plt

import sionna.rt as rt
from sionna.rt import (load_scene, scene as scene_assets,
                       Transmitter, Receiver, PlanarArray,
                       PathSolver, RadioMapSolver, Camera, RadioMaterial)

# the bundled two-room apartment scene, in scenes/ of this repository
scene = load_scene('scenes/indoor_apartment/scene.xml')
scene.frequency = 5e9
scene.tx_array = PlanarArray(num_rows=1, num_cols=1,
                             pattern='iso', polarization='V')
scene.rx_array = PlanarArray(num_rows=1, num_cols=1,
                             pattern='iso', polarization='V')

rmsolver = RadioMapSolver()
rss_dbm = None   # assigned after the radio map is computed
```

## 4. Placing the Access Point

```python
# load the custom scene
scene = load_scene('scenes/indoor_apartment/scene.xml')
scene.frequency = 5e9

# one-AP option (left room)
ap1 = Transmitter('ap1', position=[-3.5, 0.0, 2.0],
                  power_dbm=20.0)
# two-AP option (one added in the right room)
ap2 = Transmitter('ap2', position=[3.5, 0.0, 2.0],
                  power_dbm=20.0)
```

## 5. Building the Received Power Map

```python
rm = rmsolver(scene, max_depth=4,
              cell_size=[0.25, 0.25],
              center=[0.0, 0.0, 1.0],   # receive plane 1.0 m above the floor
              samples_per_tx=3_000_000)
```
