## Setup

```python
import numpy as np
import matplotlib.pyplot as plt

import sionna.rt as rt
from sionna.rt import (load_scene, scene as scene_assets,
                       Transmitter, Receiver, PlanarArray,
                       PathSolver, RadioMapSolver, Camera, RadioMaterial)

# the central Munich urban scene used in many chapters of this book
scene = load_scene(scene_assets.munich)
scene.frequency = 3.5e9
scene.tx_array = PlanarArray(num_rows=1, num_cols=1,
                             pattern='iso', polarization='V')
scene.rx_array = PlanarArray(num_rows=1, num_cols=1,
                             pattern='iso', polarization='V')
scene.add(Transmitter('tx0', position=[-30.0, 38.0, 35.0]))

solver = PathSolver()
rmsolver = RadioMapSolver()
```

## 2. The Isotropic Antenna

```python
scene.tx_array = PlanarArray(num_rows=1, num_cols=1,
                             vertical_spacing=0.5,
                             horizontal_spacing=0.5,
                             pattern='iso', polarization='V')
```

## 3. The Dipole Antenna

```python
scene.tx_array = PlanarArray(num_rows=1, num_cols=1,
                             vertical_spacing=0.5,
                             horizontal_spacing=0.5,
                             pattern='dipole', polarization='V')
```

## 4. Directional Antennas

```python
scene.tx_array = PlanarArray(num_rows=1, num_cols=1,
                             vertical_spacing=0.5,
                             horizontal_spacing=0.5,
                             pattern='tr38901', polarization='V')
```

## 7. Vertical and Horizontal Polarization

```python
# vertical polarization
ant_V = PlanarArray(num_rows=1, num_cols=1,
                    pattern='iso', polarization='V')
# horizontal polarization
ant_H = PlanarArray(num_rows=1, num_cols=1,
                    pattern='iso', polarization='H')
# orthogonal V and H pair (dual-polarized element)
ant_VH = PlanarArray(num_rows=1, num_cols=1,
                     pattern='iso', polarization='VH')

# compute with V on transmit and H on receive
scene.tx_array = ant_V
scene.rx_array = ant_H
```

## 8. Transmitter and Receiver Orientation

```python
import numpy as np
from sionna.rt import Transmitter

tx = Transmitter('tx0', position=[-30.0, 38.0, 35.0],
                 orientation=[np.pi/4, 0, 0])  # 45 deg about z (azimuth)
scene.add(tx)
```
